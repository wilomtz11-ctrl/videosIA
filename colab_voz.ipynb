## 4. Construir el video

```python
episodio = 'episodios/bolivia_mar.json'
extra = f'--referencia "{referencia}"' if referencia else ''
!python construir.py {episodio} --voz chatterbox {extra}
```

## 5. Descargar
Se bajan el video, la descripción con fuentes y los audios de voz. Guarda la carpeta `voz/`: en tu compu puedes reconstruir el video con `--voz archivos` sin volver a generar la voz.

```python
import glob, shutil
nombre = os.path.splitext(os.path.basename(episodio))[0]
shutil.make_archive(f'voz_{nombre}', 'zip', 'voz', nombre)
for f in glob.glob(f'salida/{nombre}*.mp4') + glob.glob(f'salida/{nombre}_descripcion.txt') + [f'voz_{nombre}.zip']:
    files.download(f)
```
